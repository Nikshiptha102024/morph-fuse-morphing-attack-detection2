```python
import os
from PIL import Image
from facenet_pytorch import MTCNN
import torch
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
mtcnn = MTCNN(image_size=384, margin=20, device=device, post_process=False)

src_base = "/content/drive/MyDrive/SelfMADpp_project/FRLL"  # original uncropped images
dst_base = "/content/drive/MyDrive/SelfMADpp_project/FRLL_mtcnn"

folders = ["raw", "amsl", "facemorpher", "opencv", "stylegan", "webmorph"]

for folder in folders:
    src_dir = f"{src_base}/{folder}"
    dst_dir = f"{dst_base}/{folder}"
    os.makedirs(dst_dir, exist_ok=True)

    files = os.listdir(src_dir)
    failed = 0

    for fname in tqdm(files, desc=f"MTCNN cropping {folder}"):
        try:
            img = Image.open(f"{src_dir}/{fname}").convert("RGB")
            face = mtcnn(img)  # returns a tensor, already cropped+resized to 384x384
            if face is None:
                failed += 1
                continue
            # Convert tensor back to PIL and save
            face_img = Image.fromarray(
                ((face.permute(1, 2, 0).numpy() + 1) / 2 * 255).astype("uint8")
            )
            face_img.save(f"{dst_dir}/{fname}")
        except Exception as e:
            print(f"Error on {fname}: {e}")
            failed += 1

    print(f"{folder}: {len(files) - failed}/{len(files)} succeeded")
```

```python
import timm
import torch
from PIL import Image
from torchvision import transforms
import numpy as np
from tqdm import tqdm
import os

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ViT-L/16, pretrained on ImageNet-21k + fine-tuned on ImageNet-1k (matches the paper's description)
model = timm.create_model("vit_large_patch16_384", pretrained=True, num_classes=0)
model.eval().to(device)

transform = transforms.Compose([
    transforms.Resize((384, 384)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5, 0.5, 0.5], std=[0.5, 0.5, 0.5]),
])

def extract_features(folder_path, batch_size=16):
    files = sorted(os.listdir(folder_path))
    features = []
    names = []

    for i in tqdm(range(0, len(files), batch_size), desc=f"Extracting {os.path.basename(folder_path)}"):
        batch_files = files[i:i+batch_size]
        imgs = []
        for f in batch_files:
            img = Image.open(f"{folder_path}/{f}").convert("RGB")
            imgs.append(transform(img))
        batch = torch.stack(imgs).to(device)

        with torch.no_grad():
            feats = model(batch)  # (B, 1024) CLS token / pooled output

        features.append(feats.cpu().numpy())
        names.extend(batch_files)

    return np.concatenate(features, axis=0), names

base = "/content/drive/MyDrive/SelfMADpp_project/FRLL_mtcnn"
folders = ["raw", "amsl", "facemorpher", "opencv", "stylegan", "webmorph"]

all_features = {}
all_names = {}

for folder in folders:
    feats, names = extract_features(f"{base}/{folder}")
    all_features[folder] = feats
    all_names[folder] = names
    print(f"{folder}: {feats.shape}")

# Save so we don't need to re-extract if session resets
np.savez(
    "/content/drive/MyDrive/SelfMADpp_project/vit_features.npz",
    **{f"{k}_feats": v for k, v in all_features.items()},
    **{f"{k}_names": np.array(v) for k, v in all_names.items()}
)
print("Saved all features to Drive.")
```

```python
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import roc_curve

def compute_eer(y_true, y_scores):
    fpr, tpr, thresholds = roc_curve(y_true, y_scores)
    fnr = 1 - tpr
    # EER is where FPR == FNR
    eer_idx = np.nanargmin(np.abs(fpr - fnr))
    eer = (fpr[eer_idx] + fnr[eer_idx]) / 2
    return eer * 100  # as percentage

raw_feats = all_features["raw"]  # bona fide, label 0
morph_types = ["amsl", "facemorpher", "opencv", "stylegan", "webmorph"]

results = {}

for morph_type in morph_types:
    morph_feats = all_features[morph_type]  # label 1

    X = np.concatenate([raw_feats, morph_feats], axis=0)
    y = np.concatenate([np.zeros(len(raw_feats)), np.ones(len(morph_feats))])

    # Simple train/test split (80/20) since paper trains SVM per dataset
    from sklearn.model_selection import train_test_split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, stratify=y, random_state=42
    )

    clf = SVC(kernel="linear", probability=True)
    clf.fit(X_train, y_train)

    scores = clf.predict_proba(X_test)[:, 1]  # probability of being "morph"
    eer = compute_eer(y_test, scores)

    results[morph_type] = eer
    print(f"{morph_type}: D-EER = {eer:.2f}%")

print("\nMean D-EER across all morph types:", np.mean(list(results.values())), "%")
```

```python
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import roc_curve

def compute_eer(y_true, y_scores):
    fpr, tpr, thresholds = roc_curve(y_true, y_scores)
    fnr = 1 - tpr
    eer_idx = np.nanargmin(np.abs(fpr - fnr))
    eer = (fpr[eer_idx] + fnr[eer_idx]) / 2
    return eer * 100

raw_feats = all_features["raw"]
morph_types = ["amsl", "facemorpher", "opencv", "stylegan", "webmorph"]

results_loo = {}

for held_out in morph_types:
    train_types = [m for m in morph_types if m != held_out]

    # Training set: bona fide + all morph types except the held-out one
    train_morph_feats = np.concatenate([all_features[m] for m in train_types], axis=0)
    X_train = np.concatenate([raw_feats, train_morph_feats], axis=0)
    y_train = np.concatenate([
        np.zeros(len(raw_feats)),
        np.ones(len(train_morph_feats))
    ])

    # Test set: bona fide (held-out portion) + the excluded morph type
    # Use a fresh slice of bona fide not used for training balance, or reuse all bona fide as negatives
    test_morph_feats = all_features[held_out]
    X_test = np.concatenate([raw_feats, test_morph_feats], axis=0)
    y_test = np.concatenate([
        np.zeros(len(raw_feats)),
        np.ones(len(test_morph_feats))
    ])

    clf = SVC(kernel="linear", probability=True)
    clf.fit(X_train, y_train)

    scores = clf.predict_proba(X_test)[:, 1]
    eer = compute_eer(y_test, scores)

    results_loo[held_out] = eer
    print(f"Trained on {train_types}, tested on UNSEEN '{held_out}': D-EER = {eer:.2f}%")

print("\nMean Leave-One-Out D-EER:", np.mean(list(results_loo.values())), "%")
```

```python
import pickle
import numpy as np

morph_types = ["amsl", "facemorpher", "opencv", "stylegan", "webmorph"]

selfmad_scores = {}  # filename -> score
selfmad_scores_by_type = {}

for morph_type in morph_types:
    with open(f"preds_output/FRLL_{morph_type}_preds.pkl", "rb") as f:
        preds = pickle.load(f)
    selfmad_scores_by_type[morph_type] = preds  # {"raw": {...}, "morphs": {...}}
    selfmad_scores.update(preds["raw"])
    selfmad_scores.update(preds["morphs"])

print("Total SelfMAD++ scored images:", len(selfmad_scores))
```

```python
data = np.load("/content/drive/MyDrive/SelfMADpp_project/vit_features.npz", allow_pickle=True)

all_features = {}
all_names = {}
for key in ["raw", "amsl", "facemorpher", "opencv", "stylegan", "webmorph"]:
    all_features[key] = data[f"{key}_feats"]
    all_names[key] = data[f"{key}_names"]

print("Reloaded features. Example raw shape:", all_features["raw"].shape)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import roc_curve

def compute_eer(y_true, y_scores):
    fpr, tpr, thresholds = roc_curve(y_true, y_scores)
    fnr = 1 - tpr
    eer_idx = np.nanargmin(np.abs(fpr - fnr))
    eer = (fpr[eer_idx] + fnr[eer_idx]) / 2
    return eer * 100

morph_types = ["amsl", "facemorpher", "opencv", "stylegan", "webmorph"]
raw_feats = all_features["raw"]
raw_names = all_names["raw"]

fusion_results = {}
selfmad_only_results = {}
vit_only_results = {}

for held_out in morph_types:
    train_types = [m for m in morph_types if m != held_out]

    train_morph_feats = np.concatenate([all_features[m] for m in train_types], axis=0)
    X_train = np.concatenate([raw_feats, train_morph_feats], axis=0)
    y_train = np.concatenate([np.zeros(len(raw_feats)), np.ones(len(train_morph_feats))])

    clf = SVC(kernel="linear", probability=True)
    clf.fit(X_train, y_train)

    # Test set: bona fide + held-out morph type, WITH filenames
    test_feats = np.concatenate([raw_feats, all_features[held_out]], axis=0)
    test_names = np.concatenate([raw_names, all_names[held_out]], axis=0)
    y_test = np.concatenate([np.zeros(len(raw_feats)), np.ones(len(all_features[held_out]))])

    vit_scores = clf.predict_proba(test_feats)[:, 1]

    # Get SelfMAD++ scores for the SAME images, using the per-type dict (not flattened!)
    selfmad_preds = selfmad_scores_by_type[held_out]
    selfmad_score_lookup = {**selfmad_preds["raw"], **selfmad_preds["morphs"]}

    matched_vit = []
    matched_selfmad = []
    matched_labels = []

    for name, label, vscore in zip(test_names, y_test, vit_scores):
        if name in selfmad_score_lookup:
            matched_vit.append(vscore)
            matched_selfmad.append(selfmad_score_lookup[name])
            matched_labels.append(label)

    matched_vit = np.array(matched_vit)
    matched_selfmad = np.array(matched_selfmad)
    matched_labels = np.array(matched_labels)

    fused_scores = (matched_vit + matched_selfmad) / 2

    fusion_eer = compute_eer(matched_labels, fused_scores)
    vit_eer = compute_eer(matched_labels, matched_vit)
    selfmad_eer = compute_eer(matched_labels, matched_selfmad)

    fusion_results[held_out] = fusion_eer
    vit_only_results[held_out] = vit_eer
    selfmad_only_results[held_out] = selfmad_eer

    print(f"{held_out}: matched {len(matched_labels)} images | ViT-only={vit_eer:.2f}% | SelfMAD++-only={selfmad_eer:.2f}% | FUSED={fusion_eer:.2f}%")

print("\n--- MEANS ---")
print("ViT+SVM only mean EER:", np.mean(list(vit_only_results.values())))
print("SelfMAD++ only mean EER:", np.mean(list(selfmad_only_results.values())))
print("FUSED mean EER:", np.mean(list(fusion_results.values())))
```

```python
import json

final_results = {
    "selfmad_only": selfmad_only_results,
    "vit_svm_only": vit_only_results,
    "fused": fusion_results,
    "means": {
        "selfmad_only_mean": float(np.mean(list(selfmad_only_results.values()))),
        "vit_svm_only_mean": float(np.mean(list(vit_only_results.values()))),
        "fused_mean": float(np.mean(list(fusion_results.values())))
    }
}

with open("/content/drive/MyDrive/SelfMADpp_project/final_combined_results.json", "w") as f:
    json.dump(final_results, f, indent=2)

print("Saved final results to Drive.")
print(json.dumps(final_results, indent=2))
```
